# Operators in Python
## 1. Arithematic Operators ->
### To perfrom mathematical operations on data we use arithematic operators
#### +, -, *, /, //, %, **


# Addition

```python
a = 11.5
b = 20.5
c = a+b
print(c, type(c)) 
```

# Substraction 


```python
a = 550
b = 250
c = a-b 
print(c, type(c))
```

# Multiplication 

```python
a = 22
b = 15
c = a*b
print(c, type(c))
```

# Division
## 1. Normal Division (/) -> Division of two numbers result is always in Float
## 2. Floor Division (//) -> Floor division always returns integer part of result 

```python
a = 40
b = 10
c = a/b
print(a, type(a))
print(b, type(b))
print(c, type(c))
```

```python
a = 5
b = 2 
c = a//b
print(a, type(a))
print(b, type(b))
print(c, type(c))
```

# Modulus / Modulo , Mod (%) --> 

```python
num = 100
b = 10
res1 = num//b 
res2 = num%b
print("original number = ", num)
print("After floor div by 10 = ", res1)
print("After mod by 10 = ", res2)
```
