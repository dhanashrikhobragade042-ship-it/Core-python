# Composite Assignment
### Composite assignment operator are a combination of arithematic operators and assignment operator.
#### +=, -=, *=, /=, //=, %=, **=

```python
a = 10
a = a+5
print(a)
```

```python
a = 10
a = a+5
print(a)
a += 5
print(a)
print(type(a))
a -= 5
print(a)
a *= 5
print(a)
a /= 5
print(a)
print(type(a))
a //= 5
print(a)
print(type(a))
a %= 5
print(a)
print(type(a))
a **= 5
print(a)
print(type(a))
```

# Integer :->
## We can represent integer by 4 ways 
### 1. Decimal
### 2. Binary
### 3. Octal
### 4. Hexadecimal


## 1. Decimal 
#### 0 to 9, Base 10 system

```python
a = 8
print(a)
print(type(a)) 
a = 25
print(a)
print(type(a)) 
```

## 2. Binary
#### 0 and 1, Base 2 system

```python
var = 0b1010
var2 = 0B1010
print(var)
print(type(var))
print(bin(var))
```

## 3. Octal
#### 0 to 7, Base 8 system


```python
var = 0o1010

print(var)
print(type(var))
print(oct(var))
```

## 4. Hexadecimal
#### 0 to 9, a to f, A to F


```python
var = 0xface
print(var)
print(type(var))
print(hex(var))
```
