# Operators 
## Relational / Comparison Operators 
### >, <, >=, <=

# Equality Operators 
## == (Equals to Equals), != (Not Equals to)

## Relational operators always returns boolean value (True / False) 

# 1. Greater than (>) ->

```python
john = 21
hari = 23
print(john > hari)
```

# 2. Less than (<) ->

```python
john = 21
hari = 22
print(john < hari)
```

# 3. Greater than or equal to (>=) ->

```python
john = 20
hari = 21
print(hari >= john)
```

# 4. Less than or equal to (<=) ->

```python
john = 21
hari = 22
print(john <= hari)
```

# 5. Equals to Equals (==) ->

```python
hari = 21
john = 21
print(hari == john)
```

# 6. Not Equals to (!=) ->

```python
hari = 21
john = 20
print(hari != john)
```

# Logical Operators
## and
## or
## not
### for logical operations on data we use logical operators 

# 1. and operator:-
#### If all the operands are the True then and then only output is True otherwise false. 

```python
var1 = True
var2 = True
print(var1 and var2)
```

# 2. or operator:-
#### If any operand is True then output is True.
#### If all the operands are False then and then only output is False.

```python
var1 = False
var2 = False
print(var1 or var2)
```
