### 1. Take the users age as a input and check whether the person is eligible to vote or not. 

```python
age = int(input("Enter Age: "))
if age >= 18:
    print("Eligible for vote")
else:
    print("Not Eligible for vote")
```

### 2. Take the temperature as input and check whether it is hot, normal, or cold. 

```python
temp = int(input("Enter Temperature: "))
if temp >= 100:
    print("Hot")
elif temp < 90:
    print("Normal")
else:
    print("Cold")
```

### 3. Take a number as a input and check whether the number is a multiple of 10 or not.

```python
num = int(input("Enter a number: "))
if num % 10 == 0:
    print("Multiple of 10")
else: 
    print("Not multiple of 10")
```

### 4. Take the Cost Price and Selling Price as a input and check whether there is a profit, loss, or no profit/no loss.

```python
cp = float(input("Enter Cost Price: "))
sp = float(input("Enter Selling Price: "))
if sp > cp:
    print("There is a Profit")
elif sp < cp:
    print("There is a Loss")
else:
    print("No profit, No loss")
```
