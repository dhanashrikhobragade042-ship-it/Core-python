# List Methods


### 1. append()--->
#### adds an element at the end of the list.
#### list.append(element)

```python
numbers = [10,20,30,40,50]
print(numbers)
```

```python
numbers = [10,20,30,40,50]
numbers.append(60)
print(numbers)
```

## 2. insert()--->
### adds an element at a specific position
### list.insert(index, element)

```python
numbers = [10,20,30,40,50]
numbers.insert(1 , 15)
print(numbers)
```

## 3. remove()--->
### removes a specific element from the list
### list.remove(element)

```python
numbers=[10,20,30,40,50]
numbers.remove(20)
print(numbers)
```

## 4. pop()--->
### removes an element from a specific index
### list.remove(element)

```python
numbers = [10,20,30,40,50]
numbers.pop(1)
print(numbers)
```

## 5.extend()
### adds multiple elements to the list
### list.extend(iterable)

```python
numbers = [10,20,30,40,50]
numbers.extend([60,70])
print(numbers)
```

## 6. clear()--->
### removes all elements from the list
### list.clear()

```python
numbers = [10,20,30,40,50]
numbers.clear()
print(numbers)
```

## 7. count()--->
### counts how many times an elements occurs
### list.count(element)

```python
numbers=[10,20,30,40,50]
print(numbers.count(30))
```

## 8. sort()--->
### arranges the elements in ascending order
### list.sort()

```python
numbers = [20,40,70,10]
numbers.sort()
print(numbers)
```
